```python
# AI Construction Project Risk & Delay Predictor
# Phase 1 — Dataset Exploration
# File: notebooks/01_dataset_exploration.ipynb
#
# Run each section below as a separate Jupyter cell.

# ============================================================
# CELL 1: IMPORT LIBRARIES
# ============================================================

import warnings
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

warnings.filterwarnings("ignore")

print("Libraries imported successfully.")


# ============================================================
# CELL 2: DATASET PATH
# ============================================================

DATA_PATH = Path("../data/raw")

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset folder not found: {DATA_PATH.resolve()}"
    )

print("Dataset path:", DATA_PATH.resolve())


# ============================================================
# CELL 3: FIND CSV FILES
# ============================================================

csv_files = sorted(DATA_PATH.glob("*.csv"))

print("Number of CSV files:", len(csv_files))
print()

for file in csv_files:
    print("✓", file.name)


# ============================================================
# CELL 4: LOAD ALL DATASETS
# ============================================================

datasets = {}

for file in csv_files:
    try:
        df = pd.read_csv(file)
        datasets[file.stem] = df
    except (pd.errors.EmptyDataError, pd.errors.ParserError, OSError) as e:
        print(f"❌ Error loading {file.name}: {e}")

print()
print("Datasets loaded:", len(datasets))


# ============================================================
# CELL 5: DATASET OVERVIEW
# ============================================================

print("=" * 80)
print("DATASET OVERVIEW")
print("=" * 80)

for name, df in datasets.items():
    print(
        f"{name:<35} "
        f"Rows: {len(df):>10,} | "
        f"Columns: {len(df.columns):>3}"
    )


# ============================================================
# CELL 6: REQUIRED DATASET VALIDATION
# ============================================================

required_datasets = [
    "projects",
    "activities",
    "dependencies",
    "activity_states",
    "resources",
    "resource_allocation",
    "procurement",
    "events",
    "decisions",
    "rework",
    "outcomes",
]

missing_datasets = [
    name for name in required_datasets
    if name not in datasets
]

if missing_datasets:
    print("❌ Missing datasets:")
    for name in missing_datasets:
        print("  -", name)
else:
    print("✅ All required datasets are available.")


# ============================================================
# CELL 7: PROJECT VALIDATION
# ============================================================

projects = datasets["projects"]

print("Number of projects:", projects["project_id"].nunique())

assert projects["project_id"].nunique() == 100, (
    "Expected 100 projects."
)

print("✅ Project validation passed.")


# ============================================================
# CELL 8: ACTIVITY VALIDATION
# ============================================================

activities = datasets["activities"]

print("Number of activities:", len(activities))
print(
    "Unique activity IDs:",
    activities["activity_id"].nunique()
)

assert len(activities) == 9279, (
    "Expected 9,279 activities."
)

print("✅ Activity validation passed.")


# ============================================================
# CELL 9: DEPENDENCY VALIDATION
# ============================================================

dependencies = datasets["dependencies"]

print("Number of dependencies:", len(dependencies))

assert len(dependencies) == 18176, (
    "Expected 18,176 dependencies."
)

print("✅ Dependency validation passed.")


# ============================================================
# CELL 10: FINAL DATASET VALIDATION
# ============================================================

print()
print("=" * 70)
print("DATASET VALIDATION SUMMARY")
print("=" * 70)

print("✅ Dataset validation passed")
print(
    f"✅ {projects['project_id'].nunique():,} projects, "
    f"{len(activities):,} activities, "
    f"{len(dependencies):,} dependencies loaded"
)


# ============================================================
# CELL 11: MISSING VALUE ANALYSIS
# ============================================================

quality_results = []

for name, df in datasets.items():

    missing_cells = int(df.isna().sum().sum())
    total_cells = int(df.shape[0] * df.shape[1])

    missing_percentage = (
        missing_cells / total_cells * 100
        if total_cells > 0
        else 0
    )

    duplicate_rows = int(df.duplicated().sum())

    quality_results.append({
        "dataset": name,
        "records": len(df),
        "columns": len(df.columns),
        "missing_cells": missing_cells,
        "missing_percentage": round(missing_percentage, 4),
        "duplicate_rows": duplicate_rows,
    })

quality_report = pd.DataFrame(quality_results)

display(quality_report)


# ============================================================
# CELL 12: DATA QUALITY ISSUES
# ============================================================

issues = quality_report[
    (quality_report["missing_cells"] > 0) |
    (quality_report["duplicate_rows"] > 0)
]

print("=" * 70)
print("DATA QUALITY ISSUES")
print("=" * 70)

if issues.empty:
    print("✅ No missing values or duplicate rows found.")
else:
    display(issues)


# ============================================================
# CELL 13: DETAILED MISSING VALUE ANALYSIS
# ============================================================

print("=" * 70)
print("MISSING VALUES BY DATASET")
print("=" * 70)

for name, df in datasets.items():

    missing = df.isna().sum()
    missing = missing[missing > 0]

    if len(missing) > 0:

        print()
        print(f"Dataset: {name}")

        for column, count in missing.items():
            percentage = count / len(df) * 100

            print(
                f"  {column:<35} "
                f"{count:>8,} "
                f"({percentage:.2f}%)"
            )


# ============================================================
# CELL 14: PROJECT DATASET
# ============================================================

print("=" * 70)
print("PROJECT DATASET")
print("=" * 70)

print("Shape:", projects.shape)
print()

print("Columns:")
for column in projects.columns:
    print(" -", column)


# ============================================================
# CELL 15: PROJECT STATISTICS
# ============================================================

numeric_project_columns = [
    "floors",
    "area_m2",
    "complexity",
    "contractor_capability",
    "resource_availability",
    "management_maturity",
    "weather_exposure",
    "supply_chain_exposure",
    "technology_maturity",
    "planned_duration_days",
    "planned_cost",
]

project_statistics = projects[numeric_project_columns].describe().T

display(project_statistics)


# ============================================================
# CELL 16: VISUALIZATION 1 — PROJECT DURATION
# ============================================================

plt.figure(figsize=(10, 6))

plt.hist(
    projects["planned_duration_days"],
    bins=15,
    edgecolor="black"
)

plt.title("Distribution of Planned Project Duration")
plt.xlabel("Planned Duration (Days)")
plt.ylabel("Number of Projects")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 17: VISUALIZATION 2 — PROJECT COST
# ============================================================

plt.figure(figsize=(10, 6))

plt.hist(
    projects["planned_cost"],
    bins=15,
    edgecolor="black"
)

plt.title("Distribution of Planned Project Cost")
plt.xlabel("Planned Cost")
plt.ylabel("Number of Projects")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 18: VISUALIZATION 3 — PROJECT COMPLEXITY
# ============================================================

plt.figure(figsize=(10, 6))

plt.hist(
    projects["complexity"],
    bins=15,
    edgecolor="black"
)

plt.title("Distribution of Project Complexity")
plt.xlabel("Complexity")
plt.ylabel("Number of Projects")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 19: ACTIVITY DATASET
# ============================================================

print("=" * 70)
print("ACTIVITY DATASET")
print("=" * 70)

print("Shape:", activities.shape)
print()

print("Columns:")

for column in activities.columns:
    print(" -", column)


# ============================================================
# CELL 20: ACTIVITY STATISTICS
# ============================================================

activity_numeric_columns = [
    "planned_duration_days",
    "quantity",
    "unit_cost",
    "planned_cost",
    "criticality",
    "critical_path",
    "critical_path_position",
    "project_network_duration",
    "predecessor_count",
    "successor_count",
]

activity_statistics = (
    activities[activity_numeric_columns]
    .describe()
    .T
)

display(activity_statistics)


# ============================================================
# CELL 21: VISUALIZATION 4 — RESOURCE TYPE
# ============================================================

resource_counts = (
    activities["resource_type"]
    .value_counts()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

resource_counts.plot(kind="bar")

plt.title("Activities by Resource Type")
plt.xlabel("Resource Type")
plt.ylabel("Number of Activities")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 22: VISUALIZATION 5 — ACTIVITY PHASE
# ============================================================

phase_counts = (
    activities["phase"]
    .value_counts()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

phase_counts.plot(kind="bar")

plt.title("Activities by Construction Phase")
plt.xlabel("Construction Phase")
plt.ylabel("Number of Activities")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 23: VISUALIZATION 6 — ACTIVITY DURATION
# ============================================================

plt.figure(figsize=(10, 6))

plt.hist(
    activities["planned_duration_days"],
    bins=30,
    edgecolor="black"
)

plt.title("Distribution of Activity Planned Duration")
plt.xlabel("Planned Duration (Days)")
plt.ylabel("Number of Activities")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 24: VISUALIZATION 7 — CRITICAL PATH
# ============================================================

critical_counts = (
    activities["critical_path"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(8, 6))

critical_counts.plot(kind="bar")

plt.title("Critical Path vs Non-Critical Activities")
plt.xlabel("Critical Path Flag")
plt.ylabel("Number of Activities")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# CELL 25: VISUALIZATION 8 — CRITICALITY
# ============================================================

plt.figure(figsize=(10, 6))

plt.hist(
    activities["criticality"].dropna(),
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Activity Criticality")
plt.xlabel("Criticality")
plt.ylabel("Number of Activities")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 26: VISUALIZATION 9 — DURATION VS COST
# ============================================================

sample = activities[
    [
        "planned_duration_days",
        "planned_cost"
    ]
].dropna()

plt.figure(figsize=(10, 6))

plt.scatter(
    sample["planned_duration_days"],
    sample["planned_cost"],
    alpha=0.5
)

plt.title("Activity Duration vs Planned Cost")
plt.xlabel("Planned Duration (Days)")
plt.ylabel("Planned Cost")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 27: VISUALIZATION 10 — ACTIVITIES PER PROJECT
# ============================================================

activities_per_project = (
    activities
    .groupby("project_id")
    .size()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

plt.hist(
    activities_per_project,
    bins=15,
    edgecolor="black"
)

plt.title("Distribution of Activities per Project")
plt.xlabel("Number of Activities")
plt.ylabel("Number of Projects")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 28: DEPENDENCY ANALYSIS
# ============================================================

print("=" * 70)
print("DEPENDENCY DATASET")
print("=" * 70)

print("Shape:", dependencies.shape)
print()

print("Columns:")

for column in dependencies.columns:
    print(" -", column)


# ============================================================
# CELL 29: DEPENDENCIES PER PROJECT
# ============================================================

dependency_counts = (
    dependencies
    .groupby("project_id")
    .size()
    .sort_values(ascending=False)
)

print("Average dependencies per project:")
print(round(dependency_counts.mean(), 2))

print()

print("Maximum dependencies in one project:")
print(dependency_counts.max())

print()

print("Minimum dependencies in one project:")
print(dependency_counts.min())


# ============================================================
# CELL 30: RESOURCE ANALYSIS
# ============================================================

resources = datasets["resources"]

print("=" * 70)
print("RESOURCE DATASET")
print("=" * 70)

print("Shape:", resources.shape)
print()

print("Columns:")

for column in resources.columns:
    print(" -", column)


# ============================================================
# CELL 31: RESOURCE TYPES
# ============================================================

print(resources.head())
print()

for column in resources.columns:

    if resources[column].dtype == "object":

        print(f"\nColumn: {column}")

        print(
            resources[column]
            .value_counts()
            .head(15)
        )


# ============================================================
# CELL 32: PROCUREMENT ANALYSIS
# ============================================================

procurement = datasets["procurement"]

print("=" * 70)
print("PROCUREMENT DATASET")
print("=" * 70)

print("Shape:", procurement.shape)
print()

print(procurement.head())


# ============================================================
# CELL 33: PROCUREMENT DELAY
# ============================================================

if "procurement_delay_days" in procurement.columns:

    print(
        procurement["procurement_delay_days"]
        .describe()
    )

    plt.figure(figsize=(10, 6))

    plt.hist(
        procurement["procurement_delay_days"].dropna(),
        bins=20,
        edgecolor="black"
    )

    plt.title("Distribution of Procurement Delay")
    plt.xlabel("Procurement Delay (Days)")
    plt.ylabel("Number of Records")

    plt.tight_layout()
    plt.show()

else:
    print("procurement_delay_days column not found.")


# ============================================================
# CELL 34: EVENT ANALYSIS
# ============================================================

events = datasets["events"]

print("=" * 70)
print("EVENT DATASET")
print("=" * 70)

print("Shape:", events.shape)
print()

print("Event types:")

if "event_type" in events.columns:
    print(
        events["event_type"]
        .value_counts()
    )


# ============================================================
# CELL 35: EVENT DURATION
# ============================================================

if "duration_days" in events.columns:

    plt.figure(figsize=(10, 6))

    plt.hist(
        events["duration_days"].dropna(),
        bins=20,
        edgecolor="black"
    )

    plt.title("Distribution of Event Duration")
    plt.xlabel("Event Duration (Days)")
    plt.ylabel("Number of Events")

    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 36: DECISION ANALYSIS
# ============================================================

decisions = datasets["decisions"]

print("=" * 70)
print("DECISION DATASET")
print("=" * 70)

print("Shape:", decisions.shape)
print()

if "decision_delay_days" in decisions.columns:

    print(
        decisions["decision_delay_days"]
        .describe()
    )


# ============================================================
# CELL 37: DECISION DELAY VISUALIZATION
# ============================================================

if "decision_delay_days" in decisions.columns:

    plt.figure(figsize=(10, 6))

    plt.hist(
        decisions["decision_delay_days"].dropna(),
        bins=20,
        edgecolor="black"
    )

    plt.title("Distribution of Decision Delays")
    plt.xlabel("Decision Delay (Days)")
    plt.ylabel("Number of Decisions")

    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 38: MASTER PROJECT DATASET
# ============================================================

master = datasets["master_projects_canonical"]

print("=" * 70)
print("MASTER PROJECT DATASET")
print("=" * 70)

print("Shape:", master.shape)
print()

print("Columns:")

for column in master.columns:
    print(" -", column)


# ============================================================
# CELL 39: DELAY / RISK VARIABLES
# ============================================================

delay_keywords = [
    "delay",
    "risk",
    "event",
    "decision",
    "friction",
]

delay_columns = [
    column
    for column in master.columns
    if any(
        keyword in column.lower()
        for keyword in delay_keywords
    )
]

print("Delay / risk related columns:")

for column in delay_columns:
    print(" -", column)


# ============================================================
# CELL 40: DELAY STATISTICS
# ============================================================

delay_statistics = {}

for column in delay_columns:

    if pd.api.types.is_numeric_dtype(master[column]):

        delay_statistics[column] = {
            "min": master[column].min(),
            "max": master[column].max(),
            "mean": master[column].mean(),
            "median": master[column].median(),
            "unique": master[column].nunique(),
        }

delay_statistics_df = pd.DataFrame(delay_statistics).T

display(delay_statistics_df)


# ============================================================
# CELL 41: TRAIN / VALIDATION / TEST SPLIT
# ============================================================

if "split" in projects.columns:

    split_counts = projects["split"].value_counts()

    print("=" * 70)
    print("PROJECT SPLIT")
    print("=" * 70)

    print(split_counts)

    plt.figure(figsize=(8, 6))

    split_counts.plot(kind="bar")

    plt.title("Project Train / Validation / Test Split")
    plt.xlabel("Split")
    plt.ylabel("Number of Projects")

    plt.xticks(rotation=0)

    plt.tight_layout()
    plt.show()


# ============================================================
# CELL 42: CORRELATION ANALYSIS
# ============================================================

correlation_columns = [
    "floors",
    "area_m2",
    "complexity",
    "contractor_capability",
    "resource_availability",
    "management_maturity",
    "weather_exposure",
    "supply_chain_exposure",
    "technology_maturity",
    "planned_duration_days",
    "planned_cost",
]

correlation_matrix = (
    projects[correlation_columns]
    .corr()
)

plt.figure(figsize=(12, 10))

plt.imshow(
    correlation_matrix,
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(correlation_columns)),
    correlation_columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_columns)),
    correlation_columns
)

plt.title("Project Feature Correlation Matrix")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 43: SUMMARY STATISTICS
# ============================================================

print("=" * 80)
print("PHASE 1 SUMMARY STATISTICS")
print("=" * 80)

print()

print("PROJECTS")
print("-" * 40)

print(
    "Total projects:",
    projects["project_id"].nunique()
)

print(
    "Average planned duration:",
    round(projects["planned_duration_days"].mean(), 2),
    "days"
)

print(
    "Average planned cost:",
    round(projects["planned_cost"].mean(), 2)
)

print()

print("ACTIVITIES")
print("-" * 40)

print(
    "Total activities:",
    len(activities)
)

print(
    "Average activities/project:",
    round(
        len(activities) /
        projects["project_id"].nunique(),
        2
    )
)

print(
    "Average activity duration:",
    round(
        activities["planned_duration_days"].mean(),
        2
    ),
    "days"
)

print()

print("DEPENDENCIES")
print("-" * 40)

print(
    "Total dependencies:",
    len(dependencies)
)

print(
    "Average dependencies/project:",
    round(
        len(dependencies) /
        projects["project_id"].nunique(),
        2
    )
)


# ============================================================
# CELL 44: FINAL DATA QUALITY REPORT
# ============================================================

print("=" * 80)
print("FINAL DATA QUALITY REPORT")
print("=" * 80)

print()

total_missing = quality_report["missing_cells"].sum()
total_duplicates = quality_report["duplicate_rows"].sum()

print(
    f"Total datasets analyzed: {len(quality_report)}"
)

print(
    f"Total missing cells: {total_missing:,}"
)

print(
    f"Total duplicate rows: {total_duplicates:,}"
)

print()

if total_duplicates == 0:
    print("✅ No duplicate rows detected.")
else:
    print("⚠️ Duplicate rows detected.")

if total_missing == 0:
    print("✅ No missing cells detected.")
else:
    print(
        "⚠️ Missing values detected "
        "and require review."
    )

print()

display(quality_report)


# ============================================================
# CELL 45: PHASE 1 CONCLUSION
# ============================================================

print("=" * 80)
print("PHASE 1 DATASET EXPLORATION COMPLETE")
print("=" * 80)

print()

print("DATASET")
print("-" * 40)

print("✅ Dataset validation passed")

print(
    f"✅ {projects['project_id'].nunique():,} projects, "
    f"{len(activities):,} activities, "
    f"{len(dependencies):,} dependencies loaded"
)

print()

print("VISUALIZATIONS")
print("-" * 40)

print("✅ 8+ exploratory visualizations generated")

print()

print("DATA QUALITY")
print("-" * 40)

if total_duplicates == 0:
    print("✅ No duplicate rows detected")
else:
    print("⚠️ Duplicate rows detected")

if total_missing > 0:
    print(
        "⚠️ Missing values exist in one or more datasets"
    )
else:
    print("✅ No missing values detected")

print()

print("EXPLORATION")
print("-" * 40)

print("✅ Project characteristics analyzed")
print("✅ Activity characteristics analyzed")
print("✅ Construction phases analyzed")
print("✅ Resource types analyzed")
print("✅ Critical path analyzed")
print("✅ Dependencies analyzed")
print("✅ Procurement analyzed")
print("✅ Events analyzed")
print("✅ Decision delays analyzed")
print("✅ Project splits analyzed")

print()

print("STATUS")
print("-" * 40)

print("✅ Phase 1 exploration completed")


# ============================================================
# CELL 46: RECOMMENDATIONS FOR PHASE 2–4
# ============================================================

print("=" * 80)
print("RECOMMENDATIONS FOR PHASE 2–4")
print("=" * 80)

print()

print("PHASE 2 — TARGET DEFINITION & DATA MODEL")
print("-" * 60)

print("""
1. Define the prediction target using the actual SCOPE dataset.
2. Identify the project-level variable representing delay/risk/outcome.
3. Verify prediction-time availability of every feature.
4. Prevent data leakage from future or post-completion information.
5. Create a clean project-level modeling table.
6. Preserve the existing train / validation / test project split.
""")

print()

print("PHASE 3 — CPM & SCHEDULE ANALYSIS")
print("-" * 60)

print("""
1. Validate dependency relationships.
2. Check missing predecessor references.
3. Check missing successor references.
4. Check duplicate dependency relationships.
5. Check for circular dependencies.
6. Build deterministic Critical Path Method calculations.
7. Compare calculated network duration with the dataset.
8. Test CPM using small manually verified examples.
""")

print()

print("PHASE 4 — MACHINE LEARNING BASELINES")
print("-" * 60)

print("""
1. Start with simple baseline models.
2. Logistic Regression.
3. Random Forest.
4. Gradient Boosting.
5. Evaluate Precision, Recall, F1-score, ROC-AUC and calibration.
6. Keep project-level train / validation / test separation.
7. Compare ML predictions against simple baseline rules.
8. Add explainability after establishing a reliable baseline.
""")

print()

print("=" * 80)
print("NEXT STEP")
print("=" * 80)

print(
    "➡️ After Phase 1 approval, proceed to Phase 2: "
    "Target Definition & Modeling Dataset."
)


# ============================================================
# CELL 47: EXPORT DATA QUALITY REPORT
# ============================================================

REPORT_PATH = DATA_PATH.parent.parent / "reports"

REPORT_PATH.mkdir(
    parents=True,
    exist_ok=True
)

quality_file = REPORT_PATH / "phase1_data_quality_report.csv"

quality_report.to_csv(
    quality_file,
    index=False
)

print("✅ Data quality report saved to:")
print(quality_file.resolve())


# ============================================================
# CELL 48: EXPORT SUMMARY STATISTICS
# ============================================================

summary_file = REPORT_PATH / "phase1_project_statistics.csv"

project_statistics.to_csv(
    summary_file
)

print("✅ Project statistics saved to:")
print(summary_file.resolve())


# ============================================================
# CELL 49: FINAL PHASE 1 OUTPUT
# ============================================================

print()
print("╔" + "═" * 68 + "╗")
print("║" + " " * 18 + "PHASE 1 COMPLETE" + " " * 34 + "║")
print("╠" + "═" * 68 + "╣")
print("║  ✅ Dataset validation passed" + " " * 37 + "║")
print(
    f"║  ✅ {projects['project_id'].nunique():,} projects, "
    f"{len(activities):,} activities, "
    f"{len(dependencies):,} dependencies loaded"
    + " " * 9 + "║"
)
print("║  ✅ 8+ exploratory visualizations generated" + " " * 22 + "║")
print("║  ✅ Data quality report generated" + " " * 31 + "║")
print("║  ✅ Summary statistics generated" + " " * 31 + "║")
print("║  ✅ Phase 2–4 recommendations generated" + " " * 23 + "║")
print("╚" + "═" * 68 + "╝")
```
